```python

# Assignment No. 1: Python String Methods

# Taking input from the user
name = input("Enter your name: ")
sentence = input("Enter a sentence: ")
number = input("Enter a number: ")

print("Uppercase:", name.upper())

print("Lowercase:", name.lower())

print("Title Case:", sentence.title())

print("Capitalize:", sentence.capitalize())

print("Swap Case:", sentence.swapcase())

print("Strip:", sentence.strip())

print("Left Strip:", sentence.lstrip())

print("Right Strip:", sentence.rstrip())

print("Replace:", sentence.replace("a", "@"))

print("Split:", sentence.split())

print("Join:", "-".join(sentence.split()))

print("Find:", sentence.find("a"))

print("Index:", sentence.index(sentence[0]) if sentence else -1)

print("Count:", sentence.count("a"))

print("Starts With:", name.startswith(name[0]) if name else False)

print("Ends With:", name.endswith(name[-1]) if name else False)

print("Is Alphabet:", name.isalpha())

print("Is Digit:", number.isdigit())

print("Is Alphanumeric:", name.isalnum())

print("Is Space:", sentence.isspace())

print("Is Lower:", name.islower())

print("Is Upper:", name.isupper())

print("Is Title:", sentence.istitle())

print("Center:", name.center(20, "*"))

print("Left Justify:", name.ljust(20, "-"))

print("Right Justify:", name.rjust(20, "-"))

print("Zero Fill:", number.zfill(10))

print("Partition:", sentence.partition(" "))

print("Right Partition:", sentence.rpartition(" "))

print("Casefold:", name.casefold())
```
